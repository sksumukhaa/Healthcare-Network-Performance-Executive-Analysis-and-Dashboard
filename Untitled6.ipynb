##Data Cleaning

###1. Removing duplicates

```python
duplicates = df.duplicated().sum()*100.00/ len(df)
print(f"Percentage of duplicates in df: {duplicates:.2f}%")
```

```python
df_duplicates = df[df.duplicated()]
print(f"Total duplicate rows (including all occurrences): {len(df_duplicates)}")
```

```python
#Share of duplicates is less than 1%. Thus, I'm dropping those rows
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

###2. Formatting Name column

```python
df['Name'] = df['Name'].str.strip().str.title()
```

```python
df.head(10)
```

###3. Negative billing amounts

```python
df.columns
```

```python
df.loc[df['Billing Amount'] <= 0, "Billing Amount"]
```

```python
df[df['Billing Amount'] >= 0].describe()
```

```python
#Converting these negative values to its absolute values is also a valid choice as 25%ile of entries lies within $13,241
#But, filling in with mean of billing amount (only +ve values) seems to be logical and appropriate choice

pos_billing_df = df.loc[df['Billing Amount'] >= 0, "Billing Amount"]
mean_amount = pos_billing_df.mean() #25594.63

#Fill it with mean
df['Billing Amount'] = df['Billing Amount'].apply(lambda x: mean_amount if x < 0 else x)
```

```python
df.describe()
```

###4. Format dates

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

df.dtypes
```

```python
#Calculate stay duration in days
df['Stay Duration'] = df['Discharge Date'] - df['Date of Admission']
df['Stay Duration'] = df['Stay Duration'].dt.days
```

###3. Average treatment cost change over time?

###4. Relationship between patient age and cost?

```python
df[['Age','Billing Amount']]
```

```python
df[['Age','Billing Amount']].describe()
```

```python
def get_age_group(age):
    # Calculate the lower bound of the 10-year interval
    lower = int(age // 10) * 10
    upper = lower + 10
    return f"{lower}-{upper}"

# Apply the function to create the new 'Age Group' column
df['Age Group'] = df['Age'].apply(get_age_group)

# Sort the age groups for clear representation
age_group_counts = df['Age Group'].value_counts().sort_index()
display(age_group_counts)
```

```python
amount_byage_mean = df[['Age Group', 'Billing Amount']].groupby('Age Group').mean()
amount_byage_mean
```

```python
amount_byage_mean.index
```

```python
amount_byage_mean.values
```

```python
plt.figure(figsize=(10, 6))
# Flatten the values array to 1D to avoid the TypeError
plt.bar(amount_byage_mean.index, amount_byage_mean['Billing Amount'], color="#6677BB")
plt.xlabel('Age Group')
plt.ylabel('Average Billing Amount ($)')
plt.title('Average Billing Amount per Age Group')
plt.xticks(rotation=0)
plt.show()
```
